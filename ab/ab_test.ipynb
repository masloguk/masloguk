Создадим метрики для дата-сета sample_a

```python
sample_a_prices = pd.merge(sample_a, item_prices, on=['item_id'], how='left')
sample_a_prices.head()
```

```python
#проведем горячее кодирование признака action_id
encoder = ce.OneHotEncoder(cols=['action_id'], use_cat_names=True) # указываем столбец для кодирования
type_bin = encoder.fit_transform(sample_a_prices['action_id'])
sample_a_prices = pd.concat([sample_a_prices, type_bin], axis=1)
sample_a_prices.head()
```

```python
#создадим сводную таблицу по признаку 'item_id'
a_metrics = sample_a_prices.groupby('item_id')[['action_id_1.0', 'action_id_0.0', 'action_id_2.0']].sum()
a_metrics = pd.merge(a_metrics, item_prices, on=['item_id'], how='left')
```

```python
#вычислим метрики
a_metrics['ctr'] = a_metrics['action_id_0.0'] / a_metrics['action_id_1.0']
a_metrics['purchase_rate'] = a_metrics['action_id_2.0'] / a_metrics['action_id_1.0']
a_metrics['gmv'] = a_metrics['action_id_2.0'] * a_metrics['item_price']
a_metrics.head()
```

```python
#вычислим метрики
c_metrics['ctr'] = c_metrics['action_id_0.0'] / c_metrics['action_id_1.0']
c_metrics['purchase_rate'] = c_metrics['action_id_2.0'] / c_metrics['action_id_1.0']
c_metrics['gmv'] = c_metrics['action_id_2.0'] * c_metrics['item_price']
c_metrics.head()
```

Проведем AA-тестирование. Рассмотрим метрики в группах A и С

```python
#установим уровень значимости
alpha = 0.01
#установим показатели для определения типа распределения
H0 = 'Данные распределены нормально'
Ha = 'Данные не распределены нормально (мы отвергаем H0)'
```

```python
#проведем проверку распределений на нормальность по метрике 'ctr' в сэмпле А
_, p = shapiro(a_metrics['ctr'])
print('p=%.3f' % p)
# Интерпретация 
if p > alpha:
	print(H0)
else:
	print(Ha)
```

```python
#проведем проверку распределений на нормальность по метрике 'ctr' в сэмпле C
_, p = shapiro(c_metrics['ctr'])
print('p=%.3f' % p)
# Интерпретация 
if p > alpha:
	print(H0)
else:
	print(Ha)
```

```python
#проведем проверку распределений на нормальность по метрике 'purchase_rate' в сэмпле А
_, p = shapiro(a_metrics['purchase_rate'])
print('p=%.3f' % p)
# Интерпретация 
if p > alpha:
	print(H0)
else:
	print(Ha)
```

```python
#проведем проверку распределений на нормальность по метрике 'purchase_rate' в сэмпле C
_, p = shapiro(c_metrics['purchase_rate'])
print('p=%.3f' % p)
# Интерпретация 
if p > alpha:
	print(H0)
else:
	print(Ha)
```

```python
#проведем проверку распределений на нормальность по метрике 'gmv' в сэмпле А
_, p = shapiro(a_metrics['gmv'])
print('p=%.3f' % p)
# Интерпретация 
if p > alpha:
	print(H0)
else:
	print(Ha)
```

```python
#проведем проверку распределений на нормальность по метрике 'gmv' в сэмпле C
_, p = shapiro(c_metrics['gmv'])
print('p=%.3f' % p)
# Интерпретация 
if p > alpha:
	print(H0)
else:
	print(Ha)
```

Таким образом, метрики 'ctr' и 'purchase_rate' имеют нормально распределение, а метрика 'gmv' имеет распределение, отличающееся от нормального

Для работы с метриками 'ctr' и 'purchase_rate' воспользуемся двухвыборочным т-критерием

```python
H0 = 'Нет значимой разницы между отношением кликов к просмотрам товаров.'
Ha = 'Есть значимая разница между отношением кликов к просмотрам товаров.'
```

```python
test_ctr_ac = ttest_ind(a_metrics['ctr'], c_metrics['ctr'], equal_var=True)
alpha = 0.05
p = round(test_ctr_ac[1],2)
if p>alpha:
    print(f"{p} > {alpha}. Мы не можем отвергнуть нулевую гипотезу. {H0}")
else:
    print(f"{p} <= {alpha}. Мы отвергаем нулевую гипотезу. {Ha}")
```

```python
print('Среднее значение в колонке по метрике ctr в сэмпле A: {}'.format(a_metrics['ctr'].mean()))
print('Среднее значение в колонке по метрике ctr в сэмпле C: {}'.format(c_metrics['ctr'].mean()))
```

```python
H0 = 'Нет значимой разницы между отношением покупок к просмотрам товаров.'
Ha = 'Есть значимая разница между отношением покупок к просмотрам товаров.'
```

```python
test_purchase_rate_ac = ttest_ind(a_metrics['purchase_rate'], c_metrics['purchase_rate'], equal_var=True)
alpha = 0.05
p = round(test_purchase_rate_ac[1],2)
if p>alpha:
    print(f"{p} > {alpha}. Мы не можем отвергнуть нулевую гипотезу. {H0}")
else:
    print(f"{p} <= {alpha}. Мы отвергаем нулевую гипотезу. {Ha}")
```

```python
print('Среднее значение в колонке по метрике purchase_rate в сэмпле A: {}'.format(a_metrics['purchase_rate'].mean()))
print('Среднее значение в колонке по метрике purchase_rate в сэмпле C: {}'.format(c_metrics['purchase_rate'].mean()))
```

```python
H0 = 'Нет значимой разницы в оборотах товаров.'
Ha = 'Есть значимая разница в оборотах товаров.'
```

Для анализа метрики 'gmv' воспользуемся критерием Манна-Уитни

```python
test_gmv_ac = mannwhitneyu(a_metrics['gmv'], c_metrics['gmv'], alternative='two-sided')
alpha = 0.05
p = round(test_gmv_ac[1],2)
if p>alpha:
    print(f"{p} > {alpha}. Мы не можем отвергнуть нулевую гипотезу. {H0}")
else:
    print(f"{p} <= {alpha}. Мы отвергаем нулевую гипотезу. {Ha}")
```

Поскольку есть значимая разница по всем метрикам в сэмплах А и C (при этом выборка C имеет более высокие значения по всем метрикам), то проводить сравнения между сэмплами A и B - необъективно и рисковано

Проведем AB-тестирование между сэмплами A и B

```python
#установим уровень значимости
alpha = 0.01
#установим показатели для определения типа распределения
H0 = 'Данные распределены нормально'
Ha = 'Данные не распределены нормально (мы отвергаем H0)'
```

```python
#проведем проверку распределений на нормальность по метрике 'ctr' в сэмпле B
_, p = shapiro(b_metrics['ctr'])
print('p=%.3f' % p)
# Интерпретация 
if p > alpha:
	print(H0)
else:
	print(Ha)
```

```python
#проведем проверку распределений на нормальность по метрике 'purchase_rate' в сэмпле B
_, p = shapiro(b_metrics['purchase_rate'])
print('p=%.3f' % p)
# Интерпретация 
if p > alpha:
	print(H0)
else:
	print(Ha)
```

```python
#проведем проверку распределений на нормальность по метрике 'gmv' в сэмпле B
_, p = shapiro(b_metrics['gmv'])
print('p=%.3f' % p)
# Интерпретация 
if p > alpha:
	print(H0)
else:
	print(Ha)
```

Получаем данные, которые сходятся с данными в сэмплах A и B

Для анализа метрик 'ctr' и 'purchase_rate' воспользуемся парным т-тестом

```python
H0 = 'Нет значимой разницы между отношением кликов к просмотрам товаров.'
Ha = 'Есть значимая разница между отношением кликов к просмотрам товаров.'
```

```python
test_ctr_ab = stats.ttest_rel(a_metrics['ctr'], b_metrics['ctr'])
alpha = 0.05
p = round(test_ctr_ab[1],2)
if p>alpha:
    print(f"{p} > {alpha}. Мы не можем отвергнуть нулевую гипотезу. {H0}")
else:
    print(f"{p} <= {alpha}. Мы отвергаем нулевую гипотезу. {Ha}")
```

```python
print('Среднее значение в колонке по метрике ctr в сэмпле A: {}'.format(a_metrics['ctr'].mean()))
print('Среднее значение в колонке по метрике ctr в сэмпле B: {}'.format(b_metrics['ctr'].mean()))
```

```python
H0 = 'Нет значимой разницы между отношением покупок к просмотрам товаров.'
Ha = 'Есть значимая разница между отношением покупок к просмотрам товаров.'
```

```python
test_purchase_rate_ab = stats.ttest_rel(a_metrics['purchase_rate'], b_metrics['purchase_rate'])
alpha = 0.05
p = round(test_purchase_rate_ab[1],2)
if p>alpha:
    print(f"{p} > {alpha}. Мы не можем отвергнуть нулевую гипотезу. {H0}")
else:
    print(f"{p} <= {alpha}. Мы отвергаем нулевую гипотезу. {Ha}")
```

```python
print('Среднее значение в колонке по метрике purchase_rate в сэмпле A: {}'.format(a_metrics['purchase_rate'].mean()))
print('Среднее значение в колонке по метрике purchase_rate в сэмпле B: {}'.format(b_metrics['purchase_rate'].mean()))
```
